# Dataset Metaməlumatı (Metadata)

Bu dataset istifadəçilərin e-ticarət platformasındakı davranışlarını və əməliyyatlarını əks etdirir.

## Sütunların Təsviri (Data Dictionary)

| Sütun Adı | Məlumat Tipi (Dtype) | Təsviri / İzahı | Misal Dəyər |
| :--- | :--- | :--- | :--- |
| **`event_time`** | `str` / `datetime` | Əməliyyatın və ya hadisənin baş verdiyi dəqiq tarix və vaxt (UTC formatında). | `2020-01-01 00:00:00+00:00` |
| **`event_type`** | `str` | İstifadəçi fəaliyyətinin növü (məsələn: `view` - baxış, `cart` - səbətə atma, `purchase` - alıcılıq). | `view` |
| **`product_id`** | `int64` | Məhsulun unikal identifikasiya nömrəsi (ID). | `5809910` |
| **`category_id`** | `int64` | Məhsulun aid olduğu kateqoriyanın unikal ID-si. | `1602943681873052386` |
| **`category_code`** | `str` | Kateqoriyanın oxunabilən adı/kodu (kateqoriya iyerarxiyası). Çox vaxt boş (`NaN`) ola bilər. | `appliances.environment.air_heater` / `NaN` |
| **`brand`** | `str` | Məhsulun brend/marka adı. | `grattol` |
| **`price`** | `float64` | Məhsulun qiyməti (valyuta vahidində, float tipli). | `5.24` |
| **`user_id`** | `int64` | Istifadəçinin unikal ID nömrəsi. | `595414620` |
| **`user_session`** | `str` | İstifadəçinin bir sessiyasına aid olan unikal kod (UUID). | `4adb70bb-edbd-4981-b60f-a05bfd32683a` |
